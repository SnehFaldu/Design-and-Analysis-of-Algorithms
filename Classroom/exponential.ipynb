```python
def power(x, n):
    if n == 0:
        return 1

    temp = power(x, n // 2)

    if n % 2 == 0:
        return temp * temp
    else:
        return x * temp * temp

x, n = 2, 8
print(power(x, n))
```

```python
# Merge two sorted arrays

def merging(A, B):
    merge = [0] * (len(A) + len(B))
    i = j = k = 0

    while i < len(A) and j < len(B):
        if A[i] < B[j]:
            merge[k] = A[i]
            i += 1
        else:
            merge[k] = B[j]
            j += 1
        k += 1

    while i < len(A):
        merge[k] = A[i]
        i += 1
        k += 1

    while j < len(B):
        merge[k] = B[j]
        j += 1
        k += 1

    return merge

A = [9, 10, 18, 45, 58]
B = [2, 4, 6, 8]

print(merging(A, B))
```
